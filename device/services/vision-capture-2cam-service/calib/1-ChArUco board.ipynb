```python
import cv2
import numpy as np

def generate_charuco_A3(
    squares_x=10, squares_y=7,         # จำนวนช่องแนวนอน x แนวตั้ง
    square_length_mm=30,               # ขนาดช่อง (มม.)
    marker_length_mm=18,               # ขนาดมาร์กเกอร์ (ต้อง < square_length_mm)
    dpi=300,
    out_png="charuco_A3.png"
):
    aruco = cv2.aruco
    dictionary = aruco.getPredefinedDictionary(aruco.DICT_6X6_250)

    # ✅ ใช้คอนสตรัคเตอร์แบบใหม่ (OpenCV 4.7+)
    # หมายเหตุ: อาร์กิวเมนต์ตัวแรกเป็น tuple (cols, rows)
    board = aruco.CharucoBoard(
        (squares_x, squares_y),
        square_length_mm,
        marker_length_mm,
        dictionary
    )

    # ขนาด A3 แนวนอน (มม.)
    A3_W_MM, A3_H_MM = 420.0, 297.0
    margin_mm = 10.0  # เผื่อขอบว่าง

    # แปลง มม. -> พิกเซล ตาม DPI
    mm2inch = 1/25.4
    W_px = int(A3_W_MM * mm2inch * dpi)
    H_px = int(A3_H_MM * mm2inch * dpi)
    margin_px = int(margin_mm * mm2inch * dpi)

    # ให้พื้นที่บอร์ดกินเกือบเต็มหน้า (เผื่อ margin)
    board_w_px = W_px - 2*margin_px
    board_h_px = H_px - 2*margin_px

    # ✅ วาดบอร์ดด้วย API ใหม่ (generateImage)
    # หมายเหตุ: generateImage คืนภาพ grayscale/สีตามเวอร์ชันบิลด์
    board_img = board.generateImage((board_w_px, board_h_px))

    # สร้างแคนวาสขาว A3 แล้ววางบอร์ดตรงกลาง
    if board_img.ndim == 2:
        canvas = 255 * np.ones((H_px, W_px), dtype=np.uint8)
    else:
        canvas = 255 * np.ones((H_px, W_px, 3), dtype=np.uint8)

    x = (W_px - board_w_px) // 2
    y = (H_px - board_h_px) // 2
    canvas[y:y+board_h_px, x:x+board_w_px] = board_img

    cv2.imwrite(out_png, canvas)
    print(f"✅ Saved: {out_png}  ({W_px}x{H_px}px @ {dpi}dpi)")

if __name__ == "__main__":
    generate_charuco_A3()
```
