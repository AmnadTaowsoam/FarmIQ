## capture for calib images

```python
import cv2
import numpy as np
from urllib.parse import quote
from pathlib import Path
from datetime import datetime

BASE = Path("D:/FarmIQ/device/services/vision-capture-2cam-service")
LEFT_IP = "192.168.1.199"
RIGHT_IP = "192.168.1.200"
PASSWORD = "P@ssw0rd"          # แก้หากมีการเปลี่ยน
STREAM = "/Streaming/Channels/101"
CAP_OPTS = cv2.CAP_FFMPEG      # ต้องเป็น build ที่มี FFmpeg

def grab(ip):
    url = f"rtsp://admin:{quote(PASSWORD, safe='')}@{ip}:554{STREAM}"
    cap = cv2.VideoCapture(url, CAP_OPTS)
    ok, frame = cap.read()
    cap.release()
    if not ok:
        raise RuntimeError(f"Failed to read frame from {ip}")
    return frame

def capture_pair(save=True, label="charuco"):
    left = grab(LEFT_IP)
    right = grab(RIGHT_IP)
    if not save:
        return left, right

    ts = datetime.utcnow().strftime("%Y%m%d_%H%M%S_%f")
    left_dir = BASE / "calib" / "samples" / "left"
    right_dir = BASE / "calib" / "samples" / "right"
    left_dir.mkdir(parents=True, exist_ok=True)
    right_dir.mkdir(parents=True, exist_ok=True)

    left_path = left_dir / f"{label}_{ts}.jpg"
    right_path = right_dir / f"{label}_{ts}.jpg"
    cv2.imwrite(str(left_path), left)
    cv2.imwrite(str(right_path), right)
    print(f"Saved:\n  {left_path}\n  {right_path}")
    return left, right

# ใช้
left_frame, right_frame = capture_pair(save=True, label="charuco")
```
